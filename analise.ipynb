```python
import pandas as pd
import numpy as np

df = pd.read_csv('alunos.csv') 

df['Data de Nascimento'] = pd.to_datetime(df['Data de Nascimento'], format='%d/%m/%Y')

df.info()
```

```python
df['Modalidade'].value_counts()
```

```python
# Distribuição de graduação por modalidade
crosstab_faixas = pd.crosstab(
    df['Modalidade'], 
    df['Faixa'], 
    margins=True, 
    margins_name='Total'
)
print("--- Alunos por Modalidade e Faixa ---")
print(crosstab_faixas)
```
